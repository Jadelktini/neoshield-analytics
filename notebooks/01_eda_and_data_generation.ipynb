```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chargement des données générées
df = pd.read_csv('../data_pipeline/synthetic_transactions.csv')

# 2. Aperçu et distribution des classes
print("Répartition des fraudes :")
print(df['is_fraud'].value_counts(normalize=True))

# 3. Graphique : Distribution des montants selon la fraude
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x='is_fraud', y='amount', palette='Set2')
plt.title('Distribution des montants : Normal (0) vs Fraude (1)')
plt.show()

# 4. Graphique : Vitesse de déplacement vs Fraude
plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x='speed_kmh', y='amount', hue='is_fraud', alpha=0.7)
plt.title('Détection des anomalies de vélocité (Vitesse en km/h)')
plt.show()
```
